# 模型初始化的参数使用情况

## 1. Temperature

- 0.0-0.3：需要一致性、准确性的任务（数学计算、数据提取、分类、代码生成）
- 0.5-0.7：平衡创造性和一致性（聊天、问答）
- 0.8-1.5：创造性任务（写作、头脑风暴）
- 1.5-2.0：高度创造性（诗歌、故事创作）

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(
    model='deepseek:deepseek-v4-flash',
    temperature=0,
    api_key=DEEPSEEK_API_KEY,    # type: ignore
    base_url=DEEPSEEK_BASE_URL
)

print(model.invoke("张三，男，30岁，拥有8年编程开发经验，目前在某互联网大厂担任技术专家。帮我从上文中提取数据，返回JSON格式").content)
```

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(
    model='deepseek:deepseek-v4-flash',
    temperature=1.5,
    api_key=DEEPSEEK_API_KEY,    # type: ignore
    base_url=DEEPSEEK_BASE_URL
)

print(model.invoke("请为一款极致静音的机械键盘写3个充满诗意且极具张力的广告语。").content)
```
