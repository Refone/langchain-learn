# 调用线上代码

## 调用具体模型厂商的 API

### DeepSeek

三个重要的参数:
- base_url
- model_name
- api_key

将参数信息储存在 .env 配置中

```python
import os
from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek

# 读取 .env 配置文件中的信息.
# 若有同名环境变量,以 .env 文件中优先
load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

deepseek = ChatDeepSeek(
    model='deepseek-flash',
    api_key=DEEPSEEK_API_KEY,   # type: ignore
    base_url=DEEPSEEK_BASE_URL
)

response = deepseek.invoke('你是什么模型?')
print(response)
```

## 兼容写法

使用 `ChatOpenAI`

```python
# 加载配置文件
import os
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv(override=True)

api_key = os.getenv("DEEPSEEK_API_KEY")
base_url = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = ChatOpenAI(
    model='deepseek-flash',
    api_key=api_key,    # type: ignore
    base_url=base_url,
)

# 也可以 .env 中声明如下配置, 然后自动读取
# OPENAI_API_KEY
# OPENAI_BASE_URL
model = ChatOpenAI(
    model='deepseek-flash'
)

print(model.invoke("1+1=?").content)
```

## 使用中转平台

### 使用 OpenRouter

- [https://openrouter.ai/](https://openrouter.ai/)

```python
import os
from langchain_openrouter import ChatOpenRouter
from dotenv import load_dotenv

load_dotenv(override=True)

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")

# 获取大模型
model = ChatOpenRouter(
    model='~deepseek/deepseek-flash-latest',
    api_key=OPENROUTER_API_KEY,    # type: ignore
    base_url=OPENROUTER_API_BASE,
    max_tokens=1000
)

print(model.invoke("1+1=?"))
```

## 使用 LangChain1.x 统一方式

使用 `init_chat_model()`

```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(

    # model='deepseek-flash',
    # model_provider="deepseek",

    model='deepseek:deepseek-v4-flash',
    
    api_key=DEEPSEEK_API_KEY,    # type: ignore
    base_url=DEEPSEEK_BASE_URL
)

print(str(model.invoke("一句话介绍一下你自己"))[:100])
```
